```python
records = sorted([
    f[:-4]
    for f in os.listdir(MITDB_EXTRACTED)
    if f.endswith(".hea")
])

print("Total MIT-BIH records:", len(records))
print(records)
```

```python
from collections import Counter
import wfdb
import os
import pandas as pd

# Count all annotation symbols
symbol_counts = Counter()
record_symbol_counts = {}

for record in records:
    annotation = wfdb.rdann(
        os.path.join(MITDB_EXTRACTED, record),
        "atr"
    )

    counts = Counter(annotation.symbol)

    record_symbol_counts[record] = counts
    symbol_counts.update(annotation.symbol)

# Overall symbol summary
symbol_df = pd.DataFrame(
    symbol_counts.items(),
    columns=["Symbol", "Total_Count"]
)

symbol_df = symbol_df.sort_values(
    by="Total_Count",
    ascending=False
).reset_index(drop=True)

print("Total unique annotation symbols:", len(symbol_counts))
display(symbol_df)
```

```python
# Patient grouping
patient_groups = {}

for record in records:
    if record in ["201", "202"]:
        patient_groups[record] = "patient_201_202"
    else:
        patient_groups[record] = f"patient_{record}"

print("Total recordings:", len(records))
print("Total patient groups:", len(set(patient_groups.values())))

print("\n201 and 202 grouping:")
print(patient_groups["201"])
print(patient_groups["202"])
```

```python
from collections import Counter
import pandas as pd
import os
import wfdb

# AAMI four-class mapping
AAMI_MAPPING = {
    "N": ["N", "L", "R", "e", "j"],
    "S": ["A", "a", "J", "S"],
    "V": ["V", "E"],
    "F": ["F"]
}

# Reverse mapping
SYMBOL_TO_CLASS = {}

for class_name, symbols in AAMI_MAPPING.items():
    for symbol in symbols:
        SYMBOL_TO_CLASS[symbol] = class_name

# Q class symbols (excluded from four-class classification)
Q_SYMBOLS = ["/", "f", "Q"]

total_class_counts = Counter()
q_class_counts = Counter()
other_excluded_counts = Counter()

record_class_counts = {}

for record in records:

    annotation = wfdb.rdann(
        os.path.join(MITDB_EXTRACTED, record),
        "atr"
    )

    class_counts = Counter()

    for symbol in annotation.symbol:

        if symbol in SYMBOL_TO_CLASS:
            class_name = SYMBOL_TO_CLASS[symbol]
            class_counts[class_name] += 1
            total_class_counts[class_name] += 1

        elif symbol in Q_SYMBOLS:
            q_class_counts[symbol] += 1

        else:
            other_excluded_counts[symbol] += 1

    record_class_counts[record] = class_counts

print("AAMI Classification Completed!")
```

```python
aami_df = pd.DataFrame([
    {
        "Class": c,
        "Beat_Count": total_class_counts[c]
    }
    for c in ["N", "S", "V", "F"]
])

print("AAMI Classification Summary:")
display(aami_df)

print("Total AAMI beats:",
      sum(total_class_counts.values()))

print("\nQ Class Excluded:")
print(q_class_counts)

print("\nOther Excluded Annotations:")
print(other_excluded_counts)
```

```python
record_df = pd.DataFrame.from_dict(
    record_class_counts,
    orient="index"
).fillna(0).astype(int)

record_df.index.name = "Record"

record_df = record_df.reset_index()

print("Record-wise AAMI Beat Counts:")
display(record_df)
```

```python
from sklearn.model_selection import StratifiedGroupKFold
import numpy as np
import pandas as pd

SEED = 2027

print("StratifiedGroupKFold is ready!")
```

```python
X_labels = []
y_labels = []
group_labels = []
record_labels = []

for _, row in record_df.iterrows():

    record = row["Record"]
    patient_group = row["Patient_Group"]

    for class_name in ["N", "S", "V", "F"]:

        count = int(row[class_name])

        if count > 0:

            X_labels.extend([0] * count)

            y_labels.extend([class_name] * count)

            group_labels.extend([patient_group] * count)

            record_labels.extend([record] * count)


X_labels = np.array(X_labels)
y_labels = np.array(y_labels)
group_labels = np.array(group_labels)
record_labels = np.array(record_labels)

print("Total beat labels:", len(y_labels))
print("Total groups:", len(np.unique(group_labels)))
print("Classes:", np.unique(y_labels))
```

```python
from sklearn.model_selection import StratifiedGroupKFold
from collections import Counter

class_names = ["N", "S", "V", "F"]

best_split = None
best_score = float("inf")

for seed in range(200):

    # First: 80% train+validation, 20% test
    outer_cv = StratifiedGroupKFold(
        n_splits=5,
        shuffle=True,
        random_state=seed
    )

    outer_splits = list(
        outer_cv.split(X_labels, y_labels, group_labels)
    )

    trainval_idx, test_idx = outer_splits[0]

    # Second: 75% train, 25% validation
    # of the remaining 80%
    inner_cv = StratifiedGroupKFold(
        n_splits=4,
        shuffle=True,
        random_state=seed
    )

    inner_splits = list(
        inner_cv.split(
            X_labels[trainval_idx],
            y_labels[trainval_idx],
            group_labels[trainval_idx]
        )
    )

    train_local, val_local = inner_splits[0]

    train_idx = trainval_idx[train_local]
    val_idx = trainval_idx[val_local]

    # Class counts
    train_counts = Counter(y_labels[train_idx])
    val_counts = Counter(y_labels[val_idx])
    test_counts = Counter(y_labels[test_idx])

    # Require all four classes in each split
    if not all(
        counts.get(c, 0) > 0
        for counts in [train_counts, val_counts, test_counts]
        for c in class_names
    ):
        continue

    # Target fractions
    fractions = {
        "train": 0.60,
        "val": 0.20,
        "test": 0.20
    }

    total_counts = Counter(y_labels)

    # Class distribution deviation
    score = 0

    for split_name, counts in [
        ("train", train_counts),
        ("val", val_counts),
        ("test", test_counts)
    ]:

        for c in class_names:

            actual = counts[c] / total_counts[c]
            target = fractions[split_name]

            score += abs(actual - target)

    # Dataset size deviation
    score += abs(len(train_idx) / len(y_labels) - 0.60)
    score += abs(len(val_idx) / len(y_labels) - 0.20)
    score += abs(len(test_idx) / len(y_labels) - 0.20)

    if score < best_score:

        best_score = score

        best_split = (
            train_idx.copy(),
            val_idx.copy(),
            test_idx.copy(),
            seed
        )


if best_split is None:
    raise ValueError(
        "No valid split found. Check class distribution and patient groups."
    )

train_idx, val_idx, test_idx, selected_seed = best_split

print("Best split found!")
print("Selected seed:", selected_seed)
print("Split score:", best_score)
```

```python
import numpy as np
import pandas as pd
import os

SEED = 2027

# Recreate patient groups
patient_groups = {}

for record in records:
    if record in ["201", "202"]:
        patient_groups[record] = "patient_201_202"
    else:
        patient_groups[record] = f"patient_{record}"

# Group-wise class counts
record_df["Patient_Group"] = record_df["Record"].map(patient_groups)

group_df = (
    record_df
    .groupby("Patient_Group")[["N", "S", "V", "F"]]
    .sum()
    .reset_index()
)

group_df["Total_Beats"] = group_df[
    ["N", "S", "V", "F"]
].sum(axis=1)

print("Total patient groups:", len(group_df))
print("Total beats:", group_df["Total_Beats"].sum())

display(group_df.head())
```

```python
rng = np.random.default_rng(SEED)

class_names = ["N", "S", "V", "F"]

group_counts = group_df[class_names].to_numpy(dtype=np.int64)

total_class_counts = group_counts.sum(axis=0)
total_beats = group_counts.sum()

n_groups = len(group_df)

targets = np.array([0.60, 0.20, 0.20])

best_score = float("inf")
best_assignment = None

for _ in range(20000):

    # Assign each patient group to one of three splits
    assignment = rng.choice(
        3,
        size=n_groups,
        p=targets
    )

    split_counts = np.zeros((3, 4), dtype=np.int64)

    for split_id in range(3):
        split_counts[split_id] = group_counts[
            assignment == split_id
        ].sum(axis=0)

    # Every split must contain all four classes
    if np.any(split_counts == 0):
        continue

    # Class-wise target proportions
    class_fractions = (
        split_counts / total_class_counts[None, :]
    )

    # Total beat proportions
    beat_fractions = (
        split_counts.sum(axis=1) / total_beats
    )

    # Patient group proportions
    group_fractions = np.array([
        np.mean(assignment == split_id)
        for split_id in range(3)
    ])

    # Calculate deviation from target
    class_error = np.abs(
        class_fractions - targets[:, None]
    ).sum()

    beat_error = np.abs(
        beat_fractions - targets
    ).sum()

    group_error = np.abs(
        group_fractions - targets
    ).sum()

    score = class_error + beat_error + group_error

    if score < best_score:
        best_score = score
        best_assignment = assignment.copy()


if best_assignment is None:
    raise ValueError(
        "No valid split found. Check class support across patient groups."
    )

group_df["Split"] = [
    ["Train", "Validation", "Test"][x]
    for x in best_assignment
]

print("Split generation completed!")
print("Best score:", best_score)

display(group_df[["Patient_Group", "Total_Beats", "Split"]])
```

```python
train_groups = set(
    group_df.loc[group_df["Split"] == "Train", "Patient_Group"]
)

val_groups = set(
    group_df.loc[group_df["Split"] == "Validation", "Patient_Group"]
)

test_groups = set(
    group_df.loc[group_df["Split"] == "Test", "Patient_Group"]
)

print("Train groups:", len(train_groups))
print("Validation groups:", len(val_groups))
print("Test groups:", len(test_groups))

print("\nTrain-Val overlap:", train_groups & val_groups)
print("Train-Test overlap:", train_groups & test_groups)
print("Val-Test overlap:", val_groups & test_groups)
```

```python
split_summary = []

for split_name in ["Train", "Validation", "Test"]:

    subset = group_df[group_df["Split"] == split_name]

    counts = subset[class_names].sum()

    total = counts.sum()

    split_summary.append({
        "Split": split_name,
        "Patient_Groups": len(subset),
        "Total_Beats": int(total),
        "N": int(counts["N"]),
        "S": int(counts["S"]),
        "V": int(counts["V"]),
        "F": int(counts["F"]),
        "Percentage": round(total / total_beats * 100, 2)
    })

split_df = pd.DataFrame(split_summary)

display(split_df)
```
