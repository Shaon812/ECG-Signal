```python
import os
import wfdb
import pandas as pd

audit_data = []

for filename in sorted(os.listdir(MITDB_EXTRACTED)):

    if filename.endswith(".hea"):

        record_name = filename[:-4]
        record_path = os.path.join(MITDB_EXTRACTED, record_name)

        try:
            record = wfdb.rdrecord(record_path)
            annotation = wfdb.rdann(record_path, "atr")

            audit_data.append({
                "Record": record_name,
                "Sampling_Frequency": record.fs,
                "Channels": ", ".join(record.sig_name),
                "Samples": record.sig_len,
                "Annotations": len(annotation.sample),
                "Status": "OK"
            })

        except Exception as e:
            audit_data.append({
                "Record": record_name,
                "Sampling_Frequency": None,
                "Channels": None,
                "Samples": None,
                "Annotations": None,
                "Status": str(e)
            })

audit_df = pd.DataFrame(audit_data)

print(audit_df.to_string(index=False))
print("\nTotal records:", len(audit_df))
print("\nAudit status:")
print(audit_df["Status"].value_counts())
```

```python
REPORT_DIR = os.path.join(BASE_DIR, "reports")
os.makedirs(REPORT_DIR, exist_ok=True)

audit_df.to_csv(
    os.path.join(REPORT_DIR, "mitdb_audit.csv"),
    index=False
)

print("MIT-BIH audit report saved successfully!")
```

```python
import os
import wfdb
import pandas as pd

noise_records = ["bw", "ma", "em"]
nstdb_audit = []

for name in noise_records:
    record_path = os.path.join(NSTDB_EXTRACTED, name)

    try:
        record = wfdb.rdrecord(record_path)

        nstdb_audit.append({
            "Record": name,
            "Sampling_Frequency": record.fs,
            "Samples": record.sig_len,
            "Channels": ", ".join(record.sig_name),
            "Status": "OK"
        })

    except Exception as e:
        nstdb_audit.append({
            "Record": name,
            "Sampling_Frequency": None,
            "Samples": None,
            "Channels": None,
            "Status": str(e)
        })

nstdb_audit_df = pd.DataFrame(nstdb_audit)

print(nstdb_audit_df.to_string(index=False))
```

```python
nstdb_audit_path = os.path.join(
    BASE_DIR, "reports", "nstdb_audit.csv"
)

nstdb_audit_df.to_csv(nstdb_audit_path, index=False)

print("NSTDB audit report saved successfully!")
```
